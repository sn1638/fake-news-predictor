```python
import numpy as np
import pandas as pd
import re
from nltk.corpus import stopwords # nltk = natural language toolkit
from nltk.stem.porter import PorterStemmer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
```

```python
import nltk
nltk.download('stopwords')
```

```python
print(stopwords.words('english'))
```

```python
# print(stopwords.words('english'))
df=pd.read_csv('train.csv')
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df=df.fillna("")
```

```python
df['content']=df['author']+' '+df['title']
```

```python
print(df['content'])
```

```python
#stemming is the process of reducing the words in a dataset just like acting acter actress is treated as same word may be act(removing suffix)
port_stem = PorterStemmer()
```

```python
def stemming(content):
    ans=re.sub('[^a-zA-Z]',' ',content)
    ans=ans.lower()
    ans=ans.split()
    ans=[port_stem.stem(word) for word in ans if not word in stopwords.words('english')]
    ans=' '.join(ans)
    return ans
```

```python
df['content']=df['content'].apply(stemming) 
```

```python
print(df['content'])
```

```python
x=df['content']
y=df['label']
print(x)
```

```python
print(y)
```

```python
# vectorization is used to convert textual data in numbers
vectorizer=TfidfVectorizer()
vectorizer.fit(x)
x=vectorizer.transform(x)
```

```python
print(x)
x.shape
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,train_size=0.8,stratify=y,random_state=2)
```

```python
model = LogisticRegression()
```

```python
model.fit(xtrain,ytrain)
```

```python
#for the training data
pred1=model.predict(xtrain)
```

```python
accuracy1=accuracy_score(ytrain,pred1)
print(accuracy1*100,'%')
```

evaluating the model

```python
#for the test data
pred2=model.predict(xtest)
```

```python
accuracy2=accuracy_score(ytest,pred2)
print('the accuracy of the model on test data is', accuracy2*100,'%')
```
